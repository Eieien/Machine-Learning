```python
data = pd.read_csv(os.path.join(path, os.listdir(path)[0]))
data.head()
```

# Data cleanup and preprocessing

```python
dropped_data = data.drop(columns="Student_ID")
data["AI_Ethics_Concern"].unique()
data.isna().sum()
```

```python
AI_ethics_order = {"Low": 1, "Medium" : 2, "High": 3}
dropped_data["AI_Ethics_level"] = dropped_data["AI_Ethics_Concern"].map(AI_ethics_order)
leveled_data = dropped_data.drop(columns="AI_Ethics_Concern")

dummified = pd.get_dummies(leveled_data, drop_first=True)
dummified.head()
```
